Beginner Python notebook project for managing student records and finding the topper.
using loops and dictionary

```python
students = {}
n = int(input("Enter the number of students you want to add: "))

for i in range(n):
    name = input("Enter the name of the student: ")
    marks = int(input("Enter student marks: "))
    students[name] = marks

print("\n students record!")
for name,marks in students.items():
    print(name, ":", marks)

topper = max(students, key = students.get)

print("Topper is ",topper,"with",students[topper],"marks")
```
